```python
import numpy as np
import matplotlib.pyplot as plt
from rl_missile_sim.envs.missile_sim import MissileSim

import gym
%matplotlib inline
```

```python
env = gym.make("missile_sim-v0")
```

```python
obs, done = env.reset(), False

N = 4
episode_info = {
    "t": [],
    "pN": [],
    "pE": [],
    "pNv": [],
    "pEv": [],
    "p_psi": [],
    "tN": [],
    "tE": [],
    "tNv": [],
    "tEv": [],
    "r": [],
    "l": [],
    "r_dot": [],
    "l_dot": [],
}
while not done:
    ap = N * obs[2] * obs[6]
    obs, reward, done, info = env.step(ap)

    for k, v in info.items():
        episode_info[k].append(v)
        
        
    


```

```python
fig, ax = plt.subplots(figsize=(10, 8))
ax.scatter(episode_info["t"], episode_info["r"])
# ax.scatter(episode_info["tE"], episode_info["tN"])
ax.set_ylabel("Elapsed time (s)")
ax.set_ylabel("Range (m)")
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
# ax.scatter(episode_info["pE"], episode_info["pN"], label="pursuer")
ax.scatter(episode_info["tE"], episode_info["tN"], label="target")
ax.legend()
ax.set_xlabel("+E (m)")
ax.set_ylabel("+N (m)")
```
